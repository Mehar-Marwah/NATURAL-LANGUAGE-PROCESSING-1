```python
import spacy
nlp = spacy.load("en_core_web_sm")
```

```python
#1
doc = nlp("The boys are playing in the gardens.")

for token in doc:
    print(token.text, "→", token.lemma_)
```

```python
#2
text = input("Enter a sentence: ")
doc = nlp(text)

print([token.lemma_ for token in doc])
```

```python
#3
doc = nlp("The cats chased mice near the houses.")

for token in doc:
    if token.pos_ == "NOUN":
        print(token.text, "→", token.lemma_)
```

```python
#4
doc = nlp("She was reading while her friends were eating.")

for token in doc:
    if token.pos_ == "VERB":
        print(token.text, "→", token.lemma_)
```

```python
#5
doc = nlp("The children were laughing, dancing, and singing!")

lemmas = [
    token.lemma_
    for token in doc
    if not token.is_punct and not token.is_space
]

print(lemmas)
```

```python
#6
doc = nlp("The dogs were running towards the cars.")

for token in doc:
    if token.text.lower() != token.lemma_.lower():
        print(token.text, "→", token.lemma_)
```

```python
#7
doc = nlp("The Students Were Writing Letters.")

lemmas = [
    token.lemma_.lower()
    for token in doc
    if token.is_alpha
]

print(lemmas)
```

```python
#8
doc = nlp("The children are playing with their toys in the gardens.")

lemmas = [
    token.lemma_
    for token in doc
    if not token.is_stop
    and not token.is_punct
    and not token.is_space
]

print(lemmas)
```

```python
#9
from collections import Counter

doc = nlp("The cats chase a cat. The dogs chase a dog.")

lemmas = [
    token.lemma_.lower()
    for token in doc
    if token.is_alpha
]

print(dict(Counter(lemmas)))
```

```python
#10
# Create a sample file
with open("sample.txt", "w", encoding="utf-8") as file:
    file.write("The workers were building houses.")

# Read and lemmatise its contents
with open("sample.txt", "r", encoding="utf-8") as file:
    doc = nlp(file.read())

for token in doc:
    if token.is_alpha:
        print(token.text, "→", token.lemma_)
```

LEMMITIZATION USING NLTK

```python
import nltk
from nltk.stem import WordNetLemmatizer
from nltk.tokenize import word_tokenize

nltk.download("wordnet")
nltk.download("omw-1.4")
nltk.download("punkt_tab")

lemmatizer = WordNetLemmatizer()
```

```python
#1
print(lemmatizer.lemmatize("cats"))
```

```python
#2
words = ["dogs", "cars", "children", "mice"]

for word in words:
    print(word, "→", lemmatizer.lemmatize(word))
```

```python
#3
words = ["running", "played", "eating", "went"]

for word in words:
    print(word, "→", lemmatizer.lemmatize(word, pos="v"))
```

```python
#4
words = ["better", "best", "larger", "largest"]

for word in words:
    print(word, "→", lemmatizer.lemmatize(word, pos="a"))
```

```python
#5
word = input("Enter a verb: ").strip().lower()
print("Lemma:", lemmatizer.lemmatize(word, pos="v"))
```

```python
#6
text = "The cats chased mice near the houses."
tokens = word_tokenize(text)

lemmas = [lemmatizer.lemmatize(token) for token in tokens]
print(lemmas)
```

```python
#7
word = "running"

print("As a noun:", lemmatizer.lemmatize(word, pos="n"))
print("As a verb:", lemmatizer.lemmatize(word, pos="v"))
```

```python
#8
text = "The Children Have Toys, Books, and Boxes!"
tokens = word_tokenize(text)

lemmas = [
    lemmatizer.lemmatize(token.lower())
    for token in tokens
    if token.isalpha()
]

print(lemmas)
```

```python
#9
words = ["cats", "book", "mice", "houses"]

for word in words:
    lemma = lemmatizer.lemmatize(word)
    if word != lemma:
        print(word, "→", lemma)
```

```python
#10
from collections import Counter

words = ["cats", "cat", "dogs", "dog", "cats"]
lemmas = [lemmatizer.lemmatize(word) for word in words]

print(dict(Counter(lemmas)))
```
