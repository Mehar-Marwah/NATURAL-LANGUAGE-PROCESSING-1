QUESTION-14

```python
import re

text = "Contact us at student@gmail.com"

pattern = r"[a-zA-Z0-9_%+-]+@[a-zA-Z0-9.-]+\.[a-zA-Z]{1,}"

result = re.search(pattern, text)

if result:
    print("Email found:", result.group())
else:
    print("Email not found")
```
