```python
import re
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()
```

```python
#1
words = ["playing", "played", "plays", "player"]

for word in words:
    print(word, "→", stemmer.stem(word))
```

```python
#2
words = ["reading", "writing", "walking", "running", "singing"]

for word in words:
    print(word, "→", stemmer.stem(word))
```

```python
#3
words = ["cats", "dogs", "books", "cars", "students"]

for word in words:
    print(word, "→", stemmer.stem(word))
```

```python
#4
words = ["connect", "connected", "connecting",
         "connections", "connector"]

for word in words:
    print(word, "→", stemmer.stem(word))
```

```python
#5
words = ["happy", "happier", "happiest", "happiness"]

for word in words:
    print(f"{word:12} → {stemmer.stem(word)}")
```

```python
#6
words = ["organize", "organized", "organizing",
         "organization", "organizations"]

for word in words:
    print(word, "→", stemmer.stem(word))
```

```python
#7
word = input("Enter a word: ").strip().lower()

print("Original word:", word)
print("Stem:", stemmer.stem(word))
```

```python
#8
text = "Students are learning Python and developing projects!"
words = re.findall(r"\b[a-zA-Z]+\b", text.lower())

for word in words:
    print(word, "→", stemmer.stem(word))
```

```python
#9
text = "The players are playing football and practicing regularly."
words = re.findall(r"\b[a-zA-Z]+\b", text.lower())

stemmed_words = [stemmer.stem(word) for word in words]
stemmed_text = " ".join(stemmed_words)

print("Original:", text)
print("Stemmed:", stemmed_text)
```

```python
#10
words = ["playing", "book", "students", "walk", "connected"]

for word in words:
    stem = stemmer.stem(word)

    if word != stem:
        print(word, "→", stem)
```
