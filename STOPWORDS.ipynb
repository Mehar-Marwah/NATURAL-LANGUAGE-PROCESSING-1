```python
import spacy
nlp = spacy.load("en_core_web_sm")
```

```python
# QUESTION -1
words = ["a", "the", "computer", "and", "learning"]

for word in words:
    doc = nlp(word)
    for token in doc:
        print(token.text, "->", token.is_stop)
```

```python
#QUESTION-2
text = "She is reading a book in the library."
doc = nlp(text)

for token in doc:
    print(token.text, "->", token.is_stop)
```

```python
#QUESTION-3
text = "This is an interesting example of natural language processing."
doc = nlp(text)

stopwords = []

for token in doc:
    if token.is_stop:
        stopwords.append(token.text)

print(stopwords)
```

```python
#QUESTION-4
text = "Artificial intelligence is changing the world."
doc = nlp(text)

words = []

for token in doc:
    if not token.is_stop and not token.is_punct:
        words.append(token.text)

print(words)
```

```python
#QUESTION-5
text = "The children are playing in the garden."
doc = nlp(text)

words = []

for token in doc:
    if not token.is_stop:
        words.append(token.text)

print(" ".join(words))
```

```python
#QUESTION-6
text = "He is a student and he likes to learn Python."
doc = nlp(text)

count = 0

for token in doc:
    if token.is_stop:
        count = count + 1

print("Number of stopwords:", count)
```

```python
#QUESTION-7
text = "Data science is useful for solving real problems."
doc = nlp(text)

count = 0

for token in doc:
    if not token.is_stop and not token.is_punct:
        count = count + 1

print("Number of non-stopwords:", count)
```

```python
#QUESTION-8
text = "We are learning Python because it is useful."
doc = nlp(text)

stopwords = []
non_stopwords = []

for token in doc:
    if not token.is_punct:
        if token.is_stop:
            stopwords.append(token.text)
        else:
            non_stopwords.append(token.text)

print("Stopwords:", stopwords)
print("Non-stopwords:", non_stopwords)
```

```python
#QUESTION-9
sentences = [
    "This is my first Python program.",
    "The students are working on a project.",
    "She enjoys learning about technology."
]

for sentence in sentences:
    doc = nlp(sentence)
    words = []

    for token in doc:
        if not token.is_stop and not token.is_punct:
            words.append(token.text)

    print(" ".join(words))
```

```python
#QUESTION-10
text = input("Enter a sentence: ")
doc = nlp(text)

words = []

for token in doc:
    if not token.is_stop and not token.is_punct:
        words.append(token.text)

print("Cleaned sentence:", " ".join(words))
```
