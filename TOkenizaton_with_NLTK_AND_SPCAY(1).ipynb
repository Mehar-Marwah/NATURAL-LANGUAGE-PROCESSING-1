```python
#1
import nltk
import string
from nltk.tokenize import word_tokenize, sent_tokenize, wordpunct_tokenize

nltk.download("punkt_tab")
```

```python
#2
text = "Python is a popular programming language"
print(word_tokenize(text))
```

```python
#5
text = "I bought 3 books for 500 rupees"
print(word_tokenize(text))
```

```python
#6
text = "I don't like this movie."
print(word_tokenize(text))
```

```python
#7
text = "Python is easy and powerful!"
tokens = word_tokenize(text)

print("Total tokens:", len(tokens))
```

```python
#8
text = "Python is easy and powerful!"
tokens = word_tokenize(text)
words = [token for token in tokens if token not in string.punctuation]

print("Words:", words)
print("Word count:", len(words))
```

```python
#9
text = "Python Makes NLP Easy"
tokens = word_tokenize(text)
lower_tokens = [token.lower() for token in tokens]

print(lower_tokens)
```

```python
#10
text = "Python is easy. NLTK is useful for NLP."
sentences = sent_tokenize(text)

for i, sentence in enumerate(sentences, start=1):
    print(f"Sentence {i}:", word_tokenize(sentence))
```

TOKENIZATION WITH SPACY

```python
import spacy
nlp = spacy.blank("en")
```

```python
#1
doc = nlp("Students enjoy learning Python")

for token in doc:
    print(token.i, token.text)
```

```python
#2
doc = nlp("Python3 is useful")

for token in doc:
    print(token.text, token.is_alpha)
```

```python
#3
doc = nlp("I bought 5 books and 2 pens.")
words = [token.text for token in doc if token.is_alpha]

print(words)
```

```python
#4
doc = nlp("Email info@example.com or help@example.org")
emails = [token.text for token in doc if token.like_email]

print(emails)
```

```python
#5
doc = nlp("Visit https://www.example.com for details")
urls = [token.text for token in doc if token.like_url]

print(urls)
```

```python
#6
doc = nlp("Data-driven decision-making")
print([token.text for token in doc])
```

```python
#7
doc = nlp("Hello, everyone! Ready?")
punctuation = [token.text for token in doc if token.is_punct]

print("Punctuation:", punctuation)
print("Count:", len(punctuation))
```

```python
#8
doc = nlp("She bought five notebooks and 10 pens")
numbers = [token.text for token in doc if token.like_num]

print(numbers)
```

```python
#9
doc = nlp("Learn NLP today")

for token in doc:
    print(token.text, token.idx)
```

```python
#10
doc = nlp("AI and NLP are useful in INDIA")
uppercase = [token.text for token in doc if token.is_upper]

print(uppercase)
```
