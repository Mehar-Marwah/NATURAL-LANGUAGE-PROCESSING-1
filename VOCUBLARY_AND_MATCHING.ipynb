```python
import spacy
from spacy.matcher import Matcher

nlp = spacy.load("en_core_web_sm")
```

```python
#QUESTION-1
matcher = Matcher(nlp.vocab)
pattern = [{"LOWER": "computer"}]
matcher.add("Computer", [pattern])

doc = nlp("Computer classes help us learn computer skills.")

for match_id, start, end in matcher(doc):
    print(doc[start:end].text)
```

```python
#QUESTION-2
matcher = Matcher(nlp.vocab)
pattern = [{"LOWER": "machine"}, {"LOWER": "learning"}]
matcher.add("MachineLearning", [pattern])

doc = nlp("I am studying Machine Learning today.")

for match_id, start, end in matcher(doc):
    print(doc[start:end].text)
```

```python
#QUESTION-3
matcher = Matcher(nlp.vocab)
pattern = [
    {"LOWER": "data"},
    {"IS_PUNCT": True},
    {"LOWER": "science"}
]
matcher.add("DataScience", [pattern])

doc = nlp("She joined a data-science course.")

for match_id, start, end in matcher(doc):
    print(doc[start:end].text)
```

```python
#QUESTION-4
matcher = Matcher(nlp.vocab)

pattern1 = [{"LOWER": "onlineclass"}]
pattern2 = [{"LOWER": "online"}, {"LOWER": "class"}]
pattern3 = [
    {"LOWER": "online"},
    {"IS_PUNCT": True},
    {"LOWER": "class"}
]

matcher.add("OnlineClass", [pattern1, pattern2, pattern3])

doc = nlp("An onlineclass, an online class and an online-class are available.")

for match_id, start, end in matcher(doc):
    print(doc[start:end].text)
```

```python
#QUESTION-5
matcher = Matcher(nlp.vocab)
pattern = [{"TEXT": "Python"}]
matcher.add("ExactPython", [pattern])

doc = nlp("Python python PYTHON")

for match_id, start, end in matcher(doc):
    print(doc[start:end].text)
```

```python
#QUESTION-6
matcher = Matcher(nlp.vocab)
pattern = [{"LOWER": "solar"}, {"LOWER": "power"}]
matcher.add("SolarPower", [pattern])

doc = nlp("Solar Power is clean. Many homes use solar power.")

matches = matcher(doc)
print("Total matches:", len(matches))
```

```python
#QUESTION-7
matcher = Matcher(nlp.vocab)
pattern = [{"LOWER": "artificial"}, {"LOWER": "intelligence"}]
matcher.add("AI", [pattern])

doc = nlp("Artificial Intelligence is useful.")

for match_id, start, end in matcher(doc):
    rule_name = nlp.vocab.strings[match_id]
    span = doc[start:end]
    print(rule_name, ":", span.text)
```

```python
#QUESTION-8
matcher = Matcher(nlp.vocab)
pattern = [{"LOWER": "deep"}, {"LOWER": "learning"}]
matcher.add("DeepLearning", [pattern])

doc = nlp("We study deep learning today.")

for match_id, start, end in matcher(doc):
    print("Matched text:", doc[start:end].text)
    print("Start:", start)
    print("End:", end)
```

```python
#QUESTION-9
matcher = Matcher(nlp.vocab)

pattern1 = [{"LOWER": "data"}, {"LOWER": "science"}]
pattern2 = [{"LOWER": "web"}, {"LOWER": "development"}]

matcher.add("DataScience", [pattern1])
matcher.add("WebDevelopment", [pattern2])

doc = nlp("I enjoy data science and web development.")

for match_id, start, end in matcher(doc):
    print(nlp.vocab.strings[match_id], ":", doc[start:end].text)
```
