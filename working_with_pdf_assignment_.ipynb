```python
from PyPDF2 import PdfReader
reader =PdfReader("/content/CASE BASED PASSAGES.pdf")
```

NUMBER OF PAGES

```python
print("Number of pages:", len(reader.pages))
```

EXTRACTING AND DISPLAYING PAGES FROM FIRST PAGE

```python
first_page_text = reader.pages[0].extract_text() or ""

print(first_page_text)
```

EXTRACT AND DISPLAY PAGES FROM ALL

```python
full_text = ""

for page in reader.pages:
    full_text += (page.extract_text() or "") + "\n"

print(full_text)
```

Save the extracted PDF text to a text file.

```python
with open("passages.txt", "w", encoding="utf-8") as file:
    file.write(full_text)

print("Text saved successfully")
```

Read and display the contents of the saved text file.

```python
with open("passages.txt", "r", encoding="utf-8") as file:
    content = file.read()

print(content)
```

COUNTING WORDS

```python
words = full_text.split()

print("Total number of words:", len(words))
```

Counting  how many times the word “tourism” appears, ignoring capitalisation.

```python
import re

matches = re.findall(
    r"\btourism\b",
    full_text,
    flags=re.IGNORECASE
)

print("Occurrences of tourism:", len(matches))
```

Display the page numbers containing the word “teenagers”.

```python
for page_number, page in enumerate(reader.pages, start=1):
    text = page.extract_text() or ""

    if re.search(r"\bteenagers\b", text, flags=re.IGNORECASE):
        print("Found on page:", page_number)
```

Extract and display all percentage values mentioned in the PDF.

```python
percentages = re.findall(
    r"\b\d+(?:\.\d+)?\s*%",
    full_text
)

print("Percentage values:", percentages)
```

Save all non-empty lines containing “tourism” to a separate text file.

```python
with open("tourism_lines.txt", "w", encoding="utf-8") as file:
    for line in full_text.splitlines():
        if "tourism" in line.lower():
            file.write(line.strip() + "\n")

print("Matching lines saved to tourism_lines.txt")
```
