```python
from PyPDF2 import PdfReader
```

```python

reader = PdfReader("/content/SARABNOOR.txt")

full_text = "\n".join(
    page.extract_text() or ""
    for page in reader.pages
)

if not full_text.strip():
    raise ValueError("No text was extracted. Check the PDF.")

print("Extracted words:", len(full_text.split()))
```

1. Write the extracted PDF text into case_passages.txt using write() and close the file.

```python
file = open("case_passages.txt", "w", encoding="utf-8")

file.write(full_text)

file.close()

print("PDF text saved successfully")
```

2. Write three facts from the passages into facts.txt using writelines().

```python
facts = [
    "The passage reports that 70% of teenagers spend "
    "more than three hours daily on social media.\n",

    "The passage reports that 42% feel anxious when "
    "they are not checking their phones.\n",

    "The tourism passage states that tourism accounts "
    "for about 10% of global employment.\n"
]

with open("facts.txt", "w", encoding="utf-8") as file:
    file.writelines(facts)

print("Facts saved successfully")
```

3. Read and display the entire contents of case_passages.txt using read().

```python
with open("case_passages.txt", "r", encoding="utf-8") as file:
    content = file.read()

print(content)
```

4. Display the extracted passage text line by line without extra blank lines.

```python
with open("case_passages.txt", "r", encoding="utf-8") as file:
    for line in file:
        if line.strip():
            print(line.strip())
```

5. Read and display the first two facts from facts.txt using readline().

```python
with open("facts.txt", "r", encoding="utf-8") as file:
    first_fact = file.readline()
    second_fact = file.readline()

print("First fact:", first_fact.strip())
print("Second fact:", second_fact.strip())
```

6. Read all lines from case_passages.txt into a list using readlines().

```python
with open("case_passages.txt", "r", encoding="utf-8") as file:
    lines = file.readlines()

print(lines)
```

7. Process the list of extracted lines and display those containing “tourism”, ignoring capitalisation.

```python
with open("case_passages.txt", "r", encoding="utf-8") as file:
    lines = file.readlines()

for line in lines:
    if "tourism" in line.lower():
        print(line.strip())
```

8. Append two more facts from the PDF to facts.txt without deleting its existing contents.

```python
with open("facts.txt", "a", encoding="utf-8") as file:
    file.write(
        "The passage reports that 55% of teenagers "
        "use phones during school hours.\n"
    )

    file.write(
        "The tourism passage recommends banning "
        "single-use plastics.\n"
    )

print("Two new facts added")
```

9. Ask the user for their name and one thing they learned from the PDF, then save both in response.txt.

```python
name = input("Enter your name: ")
learning = input("What did you learn from the PDF? ")

with open("response.txt", "w", encoding="utf-8") as file:
    file.write("Name: " + name + "\n")
    file.write("What I learned: " + learning + "\n")

print("Response saved successfully")
```
